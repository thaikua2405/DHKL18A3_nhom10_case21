```python
import pandas as pd
#Đọc dữ liệu file merge
import pandas as pd

merge = pd.read_csv("data_clean/merge.csv")


```

```python
# Tạo cột present_flag
# Tạo cột đánh dấu Có mặt
merge["present_flag"] = (merge["status"] == "Có mặt").astype(int)

pivot_attendance_rate = pd.pivot_table(
    merge,
    index=["student_id", "class_id"],
    values="present_flag",
    aggfunc="mean"
)

pivot_attendance_rate.rename(
    columns={"present_flag": "attendance_rate"},
    inplace=True
)

pivot_attendance_rate

```

```python
absent_df = merge[merge["status"] == "Vắng"]

pivot_absent = pd.pivot_table(
    absent_df,
    index="course_name",
    columns="weekday",
    values="student_id",
    aggfunc="count",
    fill_value=0
)

pivot_absent
```

```python
pivot_cohort = pd.pivot_table(
    merge,
    index="class_id",
    columns="class_cohort",
    values="present_flag",
    aggfunc="mean"
)

pivot_cohort
```

```python
stacked_absent = pivot_absent.stack().reset_index()
stacked_absent.columns = ["course_name", "weekday", "absent_count"]
```

```python
unstacked_absent = stacked_absent.set_index(
    ["course_name", "weekday"]
).unstack()

unstacked_absent
```
