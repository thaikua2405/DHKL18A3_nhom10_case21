```python
import pandas as pd
#Đọc dữ liệu fale đã được chuẩn hoá
stu_clean = pd.read_csv("data_clean/student_class_clean.csv")
atten_clean = pd.read_csv("data_clean/attendance_log_clean.csv")
class_clean = pd.read_csv("data_clean/class_schedule_clean.csv")
```

```python
#Thống kê số sinh viên theo khoá
sv_theo_khoa = stu_clean.groupby("class_cohort")["student_id"].count()
print(sv_theo_khoa)
```

```python
#Thống kê số lượng học phần theo môn học
lop_theo_mon = class_clean.groupby("course_name")["class_id"].count()
print(lop_theo_mon)
```

```python
#Thống kê số lượng môn học theo class id 
buoi_hoc = atten_clean.groupby("class_id").size()
print(buoi_hoc)
```

```python
#Thống kê số lượt điểm danh theo trạng thái 
trang_thai = atten_clean["status"].value_counts()
print(trang_thai)
```

```python
#Các bản ghi attendance có student_id hoặc class_id không tồn tại
loi = atten_clean[
    (~atten_clean["student_id"].isin(stu_clean["student_id"])) |
    (~atten_clean["class_id"].isin(class_clean["class_id"]))
]
if loi.empty:
    print("Không có bản ghi lỗi.")
else:
    print(loi)
```
