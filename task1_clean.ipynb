```python
import pandas as pd
#Đọc file 
students = pd.read_csv("data_raw/student_class.csv", engine="python", skip_blank_lines=True)
classes = pd.read_csv("data_raw/class_schedule.csv", engine="python", skip_blank_lines=True)
attendance = pd.read_csv("data_raw/attendance_log.csv", engine="python", skip_blank_lines=True)
```

```python
#Chuẩn hoá dữ liệu
for df in [students, classes, attendance]:
    df.columns = df.columns.str.strip().str.lower()
    for col in df.select_dtypes(include="object"):
        df[col] = df[col].str.strip()
```

```python
# Chuẩn hóa student_id và class_cohort
students["student_id"] = students["student_id"].str.strip().str.lower().str.title().str.replace("_", "", regex=False)
students["full_name"] = students["full_name"].str.title().str.replace(" ", "", regex=False)
students["class_cohort"] = students["class_cohort"].str.replace(r"[-_\s]", "", regex=True).str.lower()
```

```python
# Chuẩn hóa class_id, course_name, weekday, slot
classes["class_id"] = classes["class_id"].str.replace(r"[-_\s]", "", regex=True).str.upper()
classes["course_name"] = classes["course_name"].str.title()
classes["weekday"] = classes["weekday"].str.upper()
weekday_map = {
    "THU 2": "Monday",
    "THỨ 2": "Monday",
    "MON": "Monday",

    "THU 3": "Tuesday",
    "THỨ 3": "Tuesday",}
classes["weekday"] = classes["weekday"].replace(weekday_map)
classes["slot"] = classes["slot"].str.upper().str.strip()
slot_map = {
    "TIET 1-3": "Tiết 1–3",
    "CA SANG": "Sáng"
}
classes["slot"] = classes["slot"].replace(slot_map)
```

```python
# Chuẩn hóa class_id trong attendance
attendance["class_id"] = attendance["class_id"].str.replace(r"[-_\s]", "", regex=True).str.upper()
attendance["student_id"] = attendance["student_id"].str.lower().str.title().str.replace("_", "")
attendance["date"] = pd.to_datetime(attendance["date"], dayfirst=True, errors="coerce")
```

```python
# Chuẩn hóa status
attendance["status"] = attendance["status"].str.lower()
status_mapping = {"co mat": "Có mặt","vang": "Vắng","di muon": "Đi muộn", "nghi phep": "Nghỉ phép"}
attendance["status"] = attendance["status"].replace(status_mapping)
```

```python
# Chuẩn hóa note
attendance["note"] = attendance["note"].str.strip().str.title()
note_translation = {"On Time": "Đúng giờ", "Khong Ly Do": "Không lý do", "Traffic Jam": "Tắc đường"}
attendance["note"] = attendance["note"].replace(note_translation).fillna("0")
```

```python
#Xuất file đã được chuẩn hoá
students.to_csv("data_clean/student_class_clean.csv", index=False)
classes.to_csv("data_clean/class_schedule_clean.csv", index=False)
attendance.to_csv("data_clean/attendance_log_clean.csv", index=False)
```
